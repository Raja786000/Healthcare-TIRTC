## Project Overview
This project performs exploratory data analysis on the Doctor Visits dataset using Python. The analysis follows the tasks shown in the provided project reference material and focuses on doctor visits, illness, gender, income, reduced activity and health-insurance indicators.

## Problem Statement

Healthcare visit data contains multiple patient and visit-related variables, but raw records are difficult to interpret directly. The objective is to analyse the dataset with Python and visualise useful patterns such as illness frequency, gender distribution, income and visits, reduced activity due to illness, missing values and insurance indicators.

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns

plt.rcParams["figure.figsize"] = (8, 5)
sns.set_theme(style="whitegrid")
```

```python
# Read the dataset
df = pd.read_csv("DoctorVisits.csv")
df.head()
```

## 1. Dataset structure and column information

```python
# Display complete information about columns, non-null counts, data types and memory usage
df.info()
```

## 2. Basic descriptive statistics

```python
df.describe(include="all").T
```

## 3. Count of people by gender

```python
gender_counts = df["gender"].value_counts()
print(gender_counts)

sns.countplot(data=df, x="gender")
plt.title("Number of Male and Female Patients")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

## 4. Minimum, maximum and mean income

```python
income_min = df["income"].min()
income_max = df["income"].max()
income_mean = df["income"].mean()

print(f"Minimum income: {income_min:.2f}")
print(f"Maximum income: {income_max:.2f}")
print(f"Mean income: {income_mean:.2f}")

plt.boxplot(df["income"].dropna(), vert=True)
plt.title("Income Distribution")
plt.ylabel("Income")
plt.show()
```

## 5. Average reduced-activity days for males and females due to illness

```python
reduced_by_gender = df.groupby("gender")["reduced"].mean()
print(reduced_by_gender)

sns.barplot(x=reduced_by_gender.index, y=reduced_by_gender.values)
plt.title("Average Reduced-Activity Days by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Reduced Days")
plt.show()
```

## 6. Number of people by illness count

```python
illness_counts = df["illness"].value_counts().sort_index()
print(illness_counts)

sns.barplot(x=illness_counts.index.astype(str), y=illness_counts.values)
plt.title("Number of People by Illness Count")
plt.xlabel("Number of Illnesses")
plt.ylabel("Number of People")
plt.show()
```

## 7. Check missing values with a heatmap

```python
print("Missing values by column:")
print(df.isnull().sum())

sns.heatmap(df.isnull(), cbar=False, cmap="viridis")
plt.title("Missing Values Heatmap")
plt.xlabel("Columns")
plt.ylabel("Records")
plt.show()
```

## 8. Analyse how income affects the number of hospital visits

```python
sns.scatterplot(data=df, x="income", y="visits", alpha=0.45)
plt.title("Income vs Number of Hospital Visits")
plt.xlabel("Income")
plt.ylabel("Visits")
plt.show()

print("Correlation between income and visits:",
      df[["income", "visits"]].corr().loc["income", "visits"])
```

## 9. Count and visualise males and females affected by illness

```python
illness_gender = df.groupby("gender")["illness"].sum()
print("Total illness count by gender:")
print(illness_gender)

sns.barplot(x=illness_gender.index, y=illness_gender.values)
plt.title("Total Illness Count by Gender")
plt.xlabel("Gender")
plt.ylabel("Total Illness Count")
plt.show()
```

## 10. Health-insurance coverage

```python
insurance_cols = ["private", "freepoor", "freerepat"]

insurance_pct = {}
for col in insurance_cols:
    insurance_pct[col] = df[col].astype(str).str.lower().eq("yes").mean() * 100

insurance_result = pd.Series(insurance_pct).rename({
    "private": "Private insurance",
    "freepoor": "Government / low-income",
    "freerepat": "Government / retiree"
})
print(insurance_result.round(2))

insurance_result.plot(kind="bar")
plt.title("Percentage of People with Health Insurance Indicators")
plt.ylabel("Percentage (%)")
plt.xticks(rotation=15)
plt.show()
```

## 11. Reduced days of activity due to illness based on gender

```python
reduced_gender = df.groupby("gender")["reduced"].sum().sort_values()

print(reduced_gender)

reduced_gender.plot(kind="barh")
plt.title("Total Reduced-Activity Days Due to Illness by Gender")
plt.xlabel("Reduced Days")
plt.ylabel("Gender")
plt.show()
```
